# The leaderboard

For the projector. Run the cell below and leave it running: it refreshes every few seconds with the scores that attendees save in Step 9 of the MOSAIC notebook.
Stop it with the square **Interrupt** button, and run it again to resume.

The scores are read from the shared folder named in `config.yaml` (`leaderboard: dir:`).

```python
import os, time, warnings
import yaml
from IPython.display import clear_output
from leaderboard import show_leaderboard

warnings.filterwarnings("ignore")
SCORES = yaml.safe_load(open("config.yaml"))["leaderboard"]["dir"]
AGENT = {}                          # a reference row, for example {"random agent (best of 50)": (1.0, 317)}: copy it from Step 9's output

try:
    while True:
        clear_output(wait=True)
        show_leaderboard("mosaic", base=SCORES, reference=AGENT)
        time.sleep(5)
except KeyboardInterrupt:
    print("stopped")
```
